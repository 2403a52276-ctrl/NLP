**Import Libraries**

```python
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB

from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

**Dataset**

```python
data = {
    'text': [
        "Congratulations, you won a gift card",
        "Limited seats available, book now",
        "Earn money from home easily",
        "Hey, are we still on for dinner?",
        "Can you send me the notes?",
        "Looking forward to our trip",
        "Claim your lottery winnings today",
        "Don't forget the meeting tomorrow"
    ],
    'label': [
        1, 1, 1,   # Spam
        0, 0, 1,   # Mixed: one normal marked spam
        1, 0
    ]
}

df = pd.DataFrame(data)
print(df)
```

**Feature Extraction (Bag-of-Words)**

```python
vectorizer = CountVectorizer()
X = vectorizer.fit_transform(df['text'])

y = df['label']
```

**Data Split**

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)
```

**Train Naive Bayes Model**

```python
model = MultinomialNB()
model.fit(X_train, y_train)
```

**Model Testing**

```python
y_pred = model.predict(X_test)

print("Predictions:", y_pred)
print("Actual:", y_test.values)
```

**Model Evaluation**

```python
print("Accuracy:", accuracy_score(y_test, y_pred))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))

print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```

**Model Saving**

```python
import joblib

# Save trained model
joblib.dump(model, 'naive_bayes_model.pkl')

# Save vectorizer also (VERY IMPORTANT)
joblib.dump(vectorizer, 'vectorizer.pkl')
```

**Load Model**

```python
# Load model
loaded_model = joblib.load('naive_bayes_model.pkl')

# Load vectorizer
loaded_vectorizer = joblib.load('vectorizer.pkl')
```

**Model Deployment**

```python
text = ["Free offer now"]

# Transform using saved vectorizer
text_vec = loaded_vectorizer.transform(text)

# Predict
prediction = loaded_model.predict(text_vec)

print("Prediction:", prediction[0])
```

**Case Study - 2 : News Group Dataset**

**Import Libraries**

```python
import numpy as np
import pandas as pd

# Sklearn modules
from sklearn.datasets import fetch_20newsgroups
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB

# Evaluation metrics
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix, classification_report
```

**Load Dataset**

```python
# Load dataset (subset for binary classification)
categories = ['sci.space', 'rec.sport.hockey']
data = fetch_20newsgroups(subset='all', categories=categories)

texts = data.data
labels = data.target

# Display sample
for i in range(3):
    print("Text:", texts[i][:200])
    print("Label:", labels[i])
    print("------")
```

**Text Preprocessing**

```python
import re

def preprocess(text):
    text = text.lower()  # lowercase
    text = re.sub(r'[^a-z\s]', '', text)  # remove punctuation
    return text

texts = [preprocess(t) for t in texts]
```

**Feature Extraction**

```python
vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(texts)
```

**Data Split**

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, labels, test_size=0.2, random_state=42
)
```

**Model Training**

```python
model = MultinomialNB()
model.fit(X_train, y_train)
```

**Model Testing**

```python
y_pred = model.predict(X_test)

print("Predictions:", y_pred)
print("Actual:", y_test)
```

**Model Evaluation**

```python
print("Accuracy:", accuracy_score(y_test, y_pred))
print("Precision:", precision_score(y_test, y_pred))
print("Recall:", recall_score(y_test, y_pred))
print("F1 Score:", f1_score(y_test, y_pred))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))

print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```

**Model Saving**

```python
import joblib

# Save trained model
joblib.dump(model, 'naive_bayes_model_news.pkl')

# Save vectorizer also (VERY IMPORTANT)
joblib.dump(vectorizer, 'vectorizer_news.pkl')
```

**Load Model**

```python
# Load model
loaded_model = joblib.load('naive_bayes_model_news.pkl')

# Load vectorizer
loaded_vectorizer = joblib.load('vectorizer_news.pkl')
```

**Model Deployment**

```python
test_samples = [
    "NASA launched a new satellite into space",
    "The hockey team won the championship game",
    "3D rendering and computer graphics techniques improved"
]

# Transform
test_vec = loaded_vectorizer.transform(test_samples)

# Predict
predictions = loaded_model.predict(test_vec)

# Map labels to category names
for text, pred in zip(test_samples, predictions):
    print("\nText:", text)
    print("Predicted Category:", data.target_names[pred])
```

**Case Study - 3: SMS Spam Collection Dataset**

**Dataset**

```python
sms_data = {
    'text': [
        'Free entry in a competition to win FA Cup final tickets.',
        'Urgent! Your mobile No. 077xxxxxxxx WON a £2000 prize.',
        'Hey, how are you doing today?',
        'Congratulations, you have won a lottery ticket!',
        'Can we meet tomorrow for coffee?'
    ],
    'label': [
        'spam', 'spam', 'ham', 'spam', 'ham'
    ]
}
sms_df = pd.DataFrame(sms_data)
print(sms_df)
```

**Feature Extraction (Bag-of-Words)**

```python
sms_vectorizer = CountVectorizer()
X_sms = sms_vectorizer.fit_transform(sms_df['text'])

y_sms = sms_df['label']
```

**Data Split**

```python
X_sms_train, X_sms_test, y_sms_train, y_sms_test = train_test_split(
    X_sms, y_sms, test_size=0.2, random_state=42
)
```

**Train Naive Bayes Model**

```python
sms_model = MultinomialNB()
sms_model.fit(X_sms_train, y_sms_train)
```

**Model Testing**

```python
y_sms_pred = sms_model.predict(X_sms_test)

print("Predictions:", y_sms_pred)
print("Actual:", y_sms_test.values)
```

**Model Evaluation**

```python
print("Accuracy:", accuracy_score(y_sms_test, y_sms_pred))

print("\nConfusion Matrix:")
print(confusion_matrix(y_sms_test, y_sms_pred))

print("\nClassification Report:")
print(classification_report(y_sms_test, y_sms_pred))
```

**Model Saving**

```python
import joblib

joblib.dump(sms_model, 'naive_bayes_model_sms.pkl')
joblib.dump(sms_vectorizer, 'vectorizer_sms.pkl')
```

**Model Deployment**

```python
loaded_sms_model = joblib.load('naive_bayes_model_sms.pkl')
loaded_sms_vectorizer = joblib.load('vectorizer_sms.pkl')

new_sms_samples = [
    "Congratulations, you've won a free iPhone!",
    "Hi, how are you doing today?"
]

new_sms_vec = loaded_sms_vectorizer.transform(new_sms_samples)
predictions_sms = loaded_sms_model.predict(new_sms_vec)

for text, pred in zip(new_sms_samples, predictions_sms):
    print(f"\nText: '{text}'")
    print(f"Predicted Label: {pred}")
```

**Case Study - 4: Movie Reviews Dataset (Binary Sentiment)**

**Dataset**

```python
movie_data = {
    'text': [
        'This movie was fantastic! A true masterpiece.',
        'Absolutely dreadful. Waste of time and money.',
        'It was okay, not great but not terrible either.',
        'Loved every minute of it, highly recommend.',
        'Such a boring film, fell asleep halfway through.'
    ],
    'sentiment': [
        1, 0, 0, 1, 0
    ]
} # 1 for positive, 0 for negative/neutral
movie_df = pd.DataFrame(movie_data)
print(movie_df)
```

**Feature Extraction (TF-IDF)**

```python
from sklearn.feature_extraction.text import TfidfVectorizer

movie_vectorizer = TfidfVectorizer()
X_movie = movie_vectorizer.fit_transform(movie_df['text'])

y_movie = movie_df['sentiment']
```

**Data Split**

```python
X_movie_train, X_movie_test, y_movie_train, y_movie_test = train_test_split(
    X_movie, y_movie, test_size=0.2, random_state=42
)
```

**Train Naive Bayes Model**

```python
movie_model = MultinomialNB()
movie_model.fit(X_movie_train, y_movie_train)
```

**Model Testing**

```python
y_movie_pred = movie_model.predict(X_movie_test)

print("Predictions:", y_movie_pred)
print("Actual:", y_movie_test.values)
```

**Model Evaluation**

```python
print("Accuracy:", accuracy_score(y_movie_test, y_movie_pred))

print("\nConfusion Matrix:")
print(confusion_matrix(y_movie_test, y_movie_pred))

print("\nClassification Report:")
print(classification_report(y_movie_test, y_movie_pred))
```

**Model Saving**

```python
import joblib

joblib.dump(movie_model, 'naive_bayes_model_movie.pkl')
joblib.dump(movie_vectorizer, 'vectorizer_movie.pkl')
```

**Model Deployment**

```python
loaded_movie_model = joblib.load('naive_bayes_model_movie.pkl')
loaded_movie_vectorizer = joblib.load('vectorizer_movie.pkl')

new_movie_samples = [
    "This film was absolutely brilliant and I loved every moment.",
    "Terrible movie, a complete waste of time.",
    "It was an average film, nothing special."
]

new_movie_vec = loaded_movie_vectorizer.transform(new_movie_samples)
predictions_movie = loaded_movie_model.predict(new_movie_vec)

for text, pred in zip(new_movie_samples, predictions_movie):
    sentiment_label = "Positive" if pred == 1 else "Negative/Neutral"
    print(f"\nText: '{text}'")
    print(f"Predicted Sentiment: {sentiment_label}")
```
